```python
import os
import kagglehub

path = kagglehub.dataset_download("ritwikb3/heart-disease-statlog")
files = os.listdir(path)

csv_file = [f for f in files if f.endswith('.csv')][0]


initial_df = pd.read_csv(os.path.join(path, csv_file))
display(initial_df.head())
display(initial_df.info())
```

#### 1) В качестве СВ X взяла возраст пациентов (столбец age), в качестве СВ Y - уровень артериального давления пациентов (столбец trestbps).

Беру первые 100 строк из этих столбцов:

```python
X = "age"
Y = "trestbps"
df=initial_df[[X,Y]].head(100).copy()
df.columns = ['X', 'Y']
pd.set_option('display.max_columns', 100)
print(df.head(100).T)
```

#### 2) Выборки для X и Y (заносим значения в отдельные массивы, объем выборки n вычисляем как длину массива):

```python
sample_x = df['X'].values
sample_y = df['Y'].values
print(f"X (n={len(sample_x)}):")
print(sample_x) 

print(f"\nY (n={len(sample_y)}):")
print(sample_y)
```

#### 3) Вариационные ряды выборок X и Y (сортирую массивы по возрастанию с помощью numpy):

```python
variational_x = np.sort(sample_x)
variational_y = np.sort(sample_y)

print("Вариационный ряд X:", variational_x)
print("\nВариационный ряд Y:", variational_y)
```

#### 4) Группированные выборки для X и Y с числом интервалов k=10.

  Ширина интервалов определяется по формуле: $$w = \frac{R}{k} = \frac{{max} - {min}}{k}$$

```python
k=10
x_min=sample_x.min()
x_max=sample_x.max()
R_x = x_max - x_min
w_x = R_x / k

print(f"Rx: {R_x}, w: {w_x}")
bins_x = [x_min + i * w_x for i in range(k + 1)] 
df_x = pd.DataFrame({
    'Интервалы X': [f"[{bins_x[i]:.1f}; {bins_x[i+1]:.1f}]" for i in range(len(bins_x)-1)],
})
display(df_x)

y_min=sample_y.min()
y_max=sample_y.max()
R_y = y_max - y_min
w_y = R_y / k

print(f"Rx: {R_y}, w: {w_y}")
bins_y = [y_min + i * w_y for i in range(k + 1)]
df_y = pd.DataFrame({
    'Интервалы Y': [f"[{bins_y[i]:.1f}; {bins_y[i+1]:.1f}]" for i in range(len(bins_y)-1)]
})
display(df_y)
```

Добавляем в таблицы: 

- середины интервалов (${Z_i}$) 

- частоты (${n_i}$)

- относительные частоты ($\sum_{}{} n_i$)

- накопленные частоты ($\sum_{}{} \frac{n_i}{n}$) 

- накопленные относительные частоты (Σni/n)

- плотность относительной частоты ($\frac{n_i}{nw}$)

```python
z_i_x = [(bins_x[i] + bins_x[i+1]) / 2 for i in range(k)]
intervals_x = pd.cut(sample_x, bins=bins_x, include_lowest=True)
n_i_x = intervals_x.value_counts().sort_index()
df_x['$z_i$']=z_i_x
df_x['$n_i$']=n_i_x.values
df_x['$n_i/n$'] = df_x['$n_i$'] / 100 
df_x['$\sum n_i$'] = df_x['$n_i$'].cumsum()
df_x['$\sum n_i/n$'] = df_x['$n_i/n$'].cumsum()
df_x['$n_i/(nw)$'] = df_x['$n_i$'] / (w_x*100)
display(df_x)

z_i_y = [(bins_y[i] + bins_y[i+1]) / 2 for i in range(k)]
intervals_y = pd.cut(sample_y, bins=bins_y, include_lowest=True)
n_i_y = intervals_y.value_counts().sort_index()
df_y['$z_i$']=z_i_y
df_y['$n_i$']=n_i_y.values
df_y['$n_i/n$'] = df_y['$n_i$'] / 100
df_y['$\sum n_i$'] = df_y['$n_i$'].cumsum()
df_y['$\sum n_i/n$'] = df_y['$n_i/n$'].cumsum()
df_y['$n_i/(nw)$'] = df_y['$n_i$'] / (w_x*100)
display(df_y)
```

#### 7) Точечные оценки мат. ожидания и дисперсии определяются формулами:

По несгруппированной выборке:
* Выборочное среднее (оценка мат. ожидания):
$$\bar{x} = \frac{1}{n} \sum_{i=1}^{n} x_i$$

* Несмещенная дисперсия:
$$s^2 = \frac{n}{n-1} \left( \frac{1}{n} \sum_{i=1}^{n} x_i^2 - (\bar{x})^2 \right)$$

* Выборочное среднеквадратичное отклонение:
$$s = \sqrt{s^2}$$

По группированной выборке:

* Выборочное среднее:
$$\bar{x} = \frac{1}{n} \sum_{j=1}^{k} z_j \cdot n_j$$

* Выборочная дисперсия:
$$s^2 = \frac{n}{n-1} \left( \frac{1}{n} \sum_{j=1}^{k} z_j^2 \cdot n_j - (\bar{x})^2 \right)$$

* Стандартное отклонение:
$$s = \sqrt{s^2}$$

```python
n=100

mean_sample_x = sample_x.mean() 
mean_variational_x = 1/n * (df_x['$z_i$'] * df_x['$n_i$']).sum()
var_est_sample_x = n/(n-1) * (1/n * sum(sample_x**2) - (mean_sample_x)**2)
std_sample_x = np.sqrt(var_est_sample_x)
var_est_variational_x = n/(n-1) * (1/n * (df_x['$z_i$']**2 * df_x['$n_i$']).sum() - (mean_variational_x)**2)
std_variational_x = np.sqrt(var_est_variational_x)

print(f"Оценка математического ожидания X по исходной выборке: {mean_sample_x:.2f}")
print(f"\nОценка дисперсии X по исходной выборке: {var_est_sample_x:.2f}")
print(f"\nОценка стандартного отклонения X по исходной выборке: {std_sample_x:.2f}")
print("\n===========================================================================")
print(f"\nОценка математического ожидания X по группированной выборке: {mean_variational_x:.2f}")
print(f"\nОценка дисперсии X по группированной выборке: {var_est_variational_x:.2f}")
print(f"\nОценка стандартного отклонения X по группированной выборке: {std_variational_x:.2f}")
```

```python
mean_sample_y = sample_y.mean() 
mean_variational_y = 1/n * (df_y['$z_i$'] * df_y['$n_i$']).sum()
var_est_sample_y = n/(n-1) * (1/n * sum(sample_y**2) - (mean_sample_y)**2)
std_sample_y = np.sqrt(var_est_sample_y)
var_est_variational_y = n/(n-1) * (1/n * (df_y['$z_i$']**2 * df_y['$n_i$']).sum() - (mean_variational_y)**2)
std_variational_y = np.sqrt(var_est_variational_y)

print(f"Оценка математического ожидания Y по исходной выборке: {mean_sample_y:.2f}")
print(f"\nОценка дисперсии Y по исходной выборке: {var_est_sample_y:.2f}")
print(f"\nОценка стандартного отклонения Y по исходной выборке: {std_sample_y:.2f}")
print("\n===========================================================================")
print(f"\nОценка математического ожидания Y по группированной выборке: {mean_variational_y:.2f}")
print(f"\nОценка дисперсии Y по группированной выборке: {var_est_variational_y:.2f}")
print(f"\nОценка стандартного отклонения Y по группированной выборке: {std_variational_y:.2f}")
```

#### 8) 95% и 99% доверительные интервалы для мат. ожидания и дисперсии СВ X и Y:

Формула доверительного интервала мат. ожидания:
$$\bar{x} - t_{1-\alpha/2, n-1} \cdot \frac{s}{\sqrt{n}} < M(X) < \bar{x} + t_{1-\alpha/2, n-1} \cdot \frac{s}{\sqrt{n}}$$

$t_{1-\alpha/2, n-1}$ — квантиль распределения Стьюдента уровня $1-\alpha/2$ с $n-1$ степенями свободы.

При надежности $\gamma = 0.95$ ($\alpha = 0.05$) и $n=100$: $t \approx 1.984$.

При надежности $\gamma = 0.99$ ($\alpha = 0.01$) и $n=100$: $t \approx 2.626$.



```python
from scipy import stats

gamma_95 = 0.95
gamma_99 = 0.99

# Квантиль распределения стьюдента
t_95 = stats.t.ppf((1 + gamma_95) / 2, df=n-1)
t_99 = stats.t.ppf((1 + gamma_99) / 2, df=n-1)

print(f"Квантиль Стьюдента (95%): {t_95:.3f}")
print(f"Квантиль Стьюдента (99%): {t_99:.3f}\n")

#Для X
low_x_95 = mean_variational_x - t_95 * std_variational_x / np.sqrt(n)
high_x_95 = mean_variational_x + t_95 * std_variational_x / np.sqrt(n)

low_x_99 = mean_variational_x - t_99 * std_variational_x / np.sqrt(n)
high_x_99 = mean_variational_x + t_99 * std_variational_x / np.sqrt(n)

print(f"95% доверительный интервал мат.ожидания для СВ X: ({low_x_95:.2f} ; {high_x_95:.2f})")
print(f"99% доверительный интервал мат.ожидания для СВ X: ({low_x_99:.2f} ; {high_x_99:.2f})\n")

#Для Y
low_y_95 = mean_variational_y - t_95 * std_variational_y / np.sqrt(n)
high_y_95 = mean_variational_y + t_95 * std_variational_y / np.sqrt(n)

low_y_99 = mean_variational_y - t_99 * std_variational_y / np.sqrt(n)
high_y_99 = mean_variational_y + t_99 * std_variational_y / np.sqrt(n)

print(f"95% доверительный интервал мат.ожидания для СВ Y: ({low_y_95:.2f} ; {high_y_95:.2f})")
print(f"99% доверительный интервал мат.ожидания для СВ Y: ({low_y_99:.2f} ; {high_y_99:.2f})")
```

Формула доверительного интервала дисперсии:

$$\frac{(n-1) \cdot s^2}{\chi^2_{\text{left}}} < \sigma^2 < \frac{(n-1) \cdot s^2}{\chi^2_{\text{right}}}$$

Статистика $\chi^2$ считается по формуле:

$$\chi^2 = \frac{(n-1)s^2}{\sigma^2}$$

* $\chi^2_{\text{left}} = \chi^2_{1-\alpha/2, \, n-1}$
* $\chi^2_{\text{right}} = \chi^2_{\alpha/2, \, n-1}$

```python
chi_95_left = stats.chi2.ppf(1 - 0.05/2, df=n-1)
chi_95_right = stats.chi2.ppf(0.05/2, df=n-1)

chi_99_left = stats.chi2.ppf(1 - 0.01/2, df=n-1)
chi_99_right = stats.chi2.ppf(0.01/2, df=n-1)

print(f"Критические точки Хи-квадрат (95%): left = {chi_95_left:.1f}, right = {chi_95_right:.1f}")
print(f"Критические точки Хи-квадрат (99%): left = {chi_99_left:.1f}, right = {chi_99_right:.1f}\n")


# Для X
var_x_low_95 = (n - 1) * var_est_variational_x / chi_95_left
var_x_high_95 = (n - 1) * var_est_variational_x / chi_95_right
print(f"95% доверительный интервал дисперсии для СВ X: ({var_x_low_95:.2f}; {var_x_high_95:.2f})")

var_x_low_99 = (n - 1) * var_est_variational_x / chi_99_left
var_x_high_99 = (n - 1) * var_est_variational_x / chi_99_right
print(f"99% доверительный интервал дисперсии для СВ X:: ({var_x_low_99:.2f}; {var_x_high_99:.2f})\n")


# Для Y
var_y_low_95 = (n - 1) * var_est_variational_y / chi_95_left
var_y_high_95 = (n - 1) * var_est_variational_y / chi_95_right
print(f"95% доверительный интервал дисперсии для СВ Y: ({var_y_low_95:.2f}; {var_y_high_95:.2f})")

var_y_low_99 = (n - 1) * var_est_variational_y / chi_99_left
var_y_high_99 = (n - 1) * var_est_variational_y / chi_99_right
print(f"99% доверительный интервал дисперсии для СВ Y: ({var_y_low_99:.2f}; {var_y_high_99:.2f})")
```

#### 9) Поскольку мы предполагаем <i>нормальное</i> распределение, определим параметры теоретического закона, используя метод моментов:

Плотность нормального распределения:
$$f(x) = \frac{1}{\sigma \sqrt{2\pi}} e^{-\frac{(x-a)^2}{2\sigma^2}}$$

Определены 2 неизвестных параметра:
* $a$ — математическое ожидание;
* $\sigma$ — среднее квадратическое отклонение.

Согласно методу моментов, для нахождения точечных оценок неизвестных параметров распределения необходимо приравнять теоретические моменты соответствующим выборочным моментам.

* Начальный теоретический момент первого порядка - это математическое ожидание: $\nu_1 = M(X) = a$.
* Центральный теоретический момент второго порядка - это дисперсия: $\mu_2 = D(X) = \sigma^2$.

Приравниваем их к выборочным оценкам по группированной выборке:
$$a = \bar{x}_{гр}$$
$$\sigma^2 = s_{гр}^2 \implies \sigma = s_{гр}$$

Для нормального закона распределения оценки параметров совпадают с уже вычисленными выборочным средним и исправленным выборочным стандартным отклонением.

```python
# Параметры нормального распределения
a_param_x = mean_variational_x
sigma_param_x = std_variational_x

a_param_y = mean_variational_y  
sigma_param_y = std_variational_y

print(f"СВ X: a={a_param_x:.2f}; sigma={sigma_param_x:.2f}")
print(f"\nСВ Y: a={a_param_y:.2f}; sigma={sigma_param_y:.2f}")
```

* Плотность нормального распределения для СВ X:
$$f(x) = \frac{1}{9.17 \cdot \sqrt{2\pi}} \cdot e^{-\frac{(x - 54.76)^2}{2 \cdot 9.17^2}} = \frac{1}{9.17 \cdot \sqrt{2\pi}} \cdot e^{-\frac{(x - 54.76)^2}{168.18}}$$

* Плотность нормального распределения для СВ Y:
$$f(y) = \frac{1}{16.20 \cdot \sqrt{2\pi}} \cdot e^{-\frac{(y - 130.62)^2}{2 \cdot 16.20^2}} = \frac{1}{16.20 \cdot \sqrt{2\pi}} \cdot e^{-\frac{(y - 130.62)^2}{524.88}}$$

#### 11) Проверим гипотезу о нормальном распределении, используя критерий Пирсона ($\chi^2$) с уровнем значимости $\alpha=0.05$:

* $H_0$: Случайная величина распределена по нормальному закону с параметрами ($a, \sigma$).
* $H_1$: Случайная величина распределена по иному закону.

Вероятности $p_i$ попадания случайной величины в $i$-й интервал $(X_i, X_{i+1}]$ вычисляются через функцию распределения нормального закона:
   $$p_i = \Phi_0\left(\frac{X_{i+1} - a}{\sigma}\right) - \Phi_0\left(\frac{X_i - a}{\sigma}\right)$$
   где $\Phi_0(x)$ — функция Лапласа. Реализовано через `stats.norm.cdf`.

Наблюдаемое значение $\chi^2_{\text{набл}}$ рассчитывается по формуле:
   $$\chi^2_{\text{набл}} = \sum_{i=1}^{k} \frac{(n_i - n \cdot p_i)^2}{n \cdot p_i}$$

Число степеней свободы определяется как $v = k - m - 1$, где $k = 10$ — число интервалов, а $m = 2$ — число параметров нормального распределения ($a$ и $\sigma$). То есть: $v = 10 - 2 - 1 = 7$.

Критическое значение $\chi^2_{\text{крит}}$ находится по таблице распределения Хи-квадрат на пересечении уровня значимости $\alpha = 0.05$ и числа степеней свободы $v = 7$. Реализовано как `stats.chi2.ppf(1 - alpha, v)`.

Если $\chi^2_{\text{набл}} < \chi^2_{\text{крит}}$, то нулевая гипотеза $H_0$ принимается, значит распределение нормальное, в противном случае гипотеза $H_0$ отвергается.

```python
alpha = 0.05
n = 100
k = 10
v = k - 2 - 1

zi_col = '$z_i$'
ni_col = '$n_i$'

# СВ X
p_i_x = []
for i in range(len(bins_x) - 1):
    p = stats.norm.cdf(bins_x[i+1], loc=a_param_x, scale=sigma_param_x) - \
        stats.norm.cdf(bins_x[i], loc=a_param_x, scale=sigma_param_x)
    p_i_x.append(p)
p_i_x = np.array(p_i_x) / np.sum(p_i_x)

calc_x = pd.DataFrame({
    '$z_i$': df_x[zi_col].values,
    '$n_i$': df_x[ni_col].values,
    '$p_i$': p_i_x,
    '$n \cdot p_i$': n * p_i_x
})
calc_x['$\chi^2$'] = (calc_x['$n_i$'] - calc_x['$n \cdot p_i$'])**2 / calc_x['$n \cdot p_i$']
print("Проверка гипотезы для СВ X:")
display(calc_x)

chi_obs_x = calc_x['$\chi^2$'].sum()
chi_crit = stats.chi2.ppf(1 - alpha, v)

print(f"  Результаты для СВ X:")
print(f"  Наблюдаемое значение χ²: {chi_obs_x:.4f}")
print(f"  Критическое значение χ²: {chi_crit:.4f}")
print(f"  Степеней свободы v: {v}")

if chi_obs_x < chi_crit:
    print("  Вывод: наблюдаемое значение χ² < критического значения χ². Гипотеза о нормальном распределении (H0) СВ X принимается.\n")
else:
    print("  Вывод: наблюдаемое значение χ² > критического значения χ². Гипотеза о нормальном распределении (H0) СВ X отклоняется.\n")
print("==============================================================================================================================\n")

# СВ Y
p_i_y = []
for i in range(len(bins_y) - 1):
    p = stats.norm.cdf(bins_y[i+1], loc=a_param_y, scale=sigma_param_y) - \
        stats.norm.cdf(bins_y[i], loc=a_param_y, scale=sigma_param_y)
    p_i_y.append(p)

p_i_y = np.array(p_i_y) / np.sum(p_i_y)
calc_y = pd.DataFrame({
    '$z_i$': df_y[zi_col].values,
    '$n_i$': df_y[ni_col].values,
    '$p_i$': p_i_y,
    '$n \cdot p_i$': n * p_i_y
})

calc_y['$\chi^2$'] = (calc_y['$n_i$'] - calc_y['$n \cdot p_i$'])**2 / calc_y['$n \cdot p_i$']

print("Проверка гипотезы для СВ Y:")
display(calc_y)

chi_obs_y = calc_y['$\chi^2$'].sum()

print(f"  Результаты для СВ Y:")
print(f"  Наблюдаемое значение χ²: {chi_obs_y:.4f}")
print(f"  Критическое значение χ²: {chi_crit:.4f}")
print(f"  Степеней свободы v: {v}")

if chi_obs_y < chi_crit:
    print("  Вывод: наблюдаемое значение χ² < критического значения χ². Гипотеза о нормальном распределении (H0) СВ Y принимается.")
else:
    print("  Вывод: наблюдаемое значение χ² > критического значения χ². Гипотеза о нормальном распределении (H0) СВ Y отклоняется.")
```

#### 12) Найдем параметры a и b уравнения линейной среднеквадратической регрессии методом наименьших квадратов:

Уравнение линейной среднеквадратической регрессии:
$$y = ax + b$$

Метод наименьших квадратов заключается в подборе параметров $a$ и $b$ так, чтобы минимизировать сумму квадратов отклонений фактических значений от теоретических, полученных по уравнению регрессии:
$$S(a, b) = \sum_{i=1}^{n} (y_i - (ax_i + b))^2 \longrightarrow \min$$

Для нахождения минимума функции $S(a, b)$ вычисляются частные производные по переменным $a$ и $b$ и приравниваются к нулю, в результате чего формируется система уравнений:
$$\begin{cases}
a \cdot \sum_{i=1}^{n} x_i^2 + b \cdot \sum_{i=1}^{n} x_i = \sum_{i=1}^{n} x_i y_i \\
a \cdot \sum_{i=1}^{n} x_i + b \cdot n = \sum_{i=1}^{n} y_i
\end{cases}$$

Разделив каждое уравнение на объем выборки $n$, параметры выражаются через выборочное:
$$\begin{cases}
a \cdot \overline{x^2} + b \cdot \bar{x} = \overline{xy} \\
a \cdot \bar{x} + b = \bar{y}
\end{cases}$$

Решая данную систему относительно неизвестных парметров $a$ и $b$, получаем окончательные расчетные формулы:
$$a = \frac{\overline{xy} - \bar{x} \cdot \bar{y}}{\overline{x^2} - (\bar{x})^2}$$

$$b = \bar{y} - a \cdot \bar{x}$$


```python
x_mean = sample_x.mean()
y_mean = sample_y.mean()
xy_mean = (sample_x * sample_y).mean()
x_sq_mean = (sample_x**2).mean()

a_regr = (xy_mean - x_mean * y_mean) / (x_sq_mean - x_mean**2)
b_regr = y_mean - a_regr * x_mean

print("Параметры линейной среднеквадратической регрессии Y на X:")
print(f"  a = {a_regr:.4f}")
print(f"  b = {b_regr:.4f}")

print(f"\nУравнение линейной регрессии: y = {a_regr:.4f}·x + {abs(b_regr):.4f}")
```

#### 13) Найдем коэффициенты детерминации и корреляции:

```python
covariance_xy = xy_mean - x_mean * y_mean
r_pearson = covariance_xy / (std_sample_x * std_sample_y)

r_sq = r_pearson ** 2

print(f"Коэффициент корреляции Пирсона: {r_pearson:.4f}")
print(f"Коэффициент детерминации: {r_sq:.4f}")
```

Значение  коэффициента корреляции Пирсона $r = 0.1860$ говорит о прямой слабой линейной зависимости между СВ $X$ и $Y$ $(0 < |r| < 0.3)$. 

Коэффициент детерминации составляет $R^2 = 0.0346$. $3.46%$ общей выборочной дисперсии переменной $Y$ объясняется влиянием  фактора $X$ в рамках составленной регрессионной модели, т.е. модель обладает низкой прогностической силой.

#### 14) Проверка значимости регрессии по критерию Фишера:

* $H_0: R^2 = 0$ - уравнение регрессии статистически незначимо, линейная связь между переменными отсутствует.
* $H_1: R^2 \neq 0$ — уравнение регрессии статистически значимо, модель адекватно описывает зависимость.

Наблюдаемое значение $F$-критерия вычисляется через коэффициент детерминации $R^2$:
$$F_{\text{набл}} = \frac{R^2 / d_1}{(1 - R^2) / d_2} = \frac{R^2}{1 - R^2} \cdot \frac{n - 2}{1}$$

Где число степеней свободы составляет:
* $d_1 = 1$ (для числителя - количество факторов модели);
* $d_2 = n - 2 = 98$ (для знаменателя - число степеней свободы остаточной дисперсии).

Критическое значение $F_{\text{крит}}$ определяется по таблицам F-распределения для заданного уровня значимости $\alpha = 0.05$ и степеней свободы $(1, 98)$. Реализовано как `f.ppf(1 - alpha, df1, df2)`.


```python
from scipy.stats import f
alpha = 0.05
df1 = 1
df2 = n - 2

f_obs = (r_sq / df1) / ((1 - r_sq) / df2)
f_crit = f.ppf(1 - alpha, df1, df2)

print(f"Наблюдаемое значение F-критерия: {f_obs:.4f}")
print(f"Критическое значение F-критерия: {f_crit:.4f}")
```

Наблюдаемое значение $F_{\text{набл}}$ меньше критического $F_{\text{крит}}$ ($3.5130 < 3.9381$), нулевая гипотеза $H_0$ принимается.
Уравнение регрессии статистически незначимое, модель не подходит для прогнозирования.
